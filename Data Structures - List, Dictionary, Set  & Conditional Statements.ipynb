### LIST

```python
## List (Creation, Modification and Access):

## 1. List Creation:

age_list = [20,22,25,26,27]
print(age_list)

name_list =["Veni","Sathya","Deva","Thivi","Viswa"]
print("Name :",name_list)

## 2. List Operations / Modifications:

## APPEND
name_list.append("Yazhini")
print("Append :",name_list)

## INSERT
age_list.insert(2,30)
print("Insert :", age_list)

## REMOVE 
name_list.remove("Yazhini")
print("Remove :" ,name_list)

## POP
age_list.pop()
print("Pop :",age_list)

## EXTEND
age_list.extend([29, 30, 26])
print("Extend :",age_list)

## SORT
age_list.sort(reverse = "true" )
print("Sort :", age_list)

## MAX, MIN, AND SUM OF ALL AGE 
max_age = max(age_list)
print("Max :", max_age)

min_age = min(age_list)
print("Min :", min_age)

total_age = sum(age_list)
print("Sum og ages:", total_age)

## 3. Accessing List Elements:

##INDEXING AND SLICING

print("First name:",name_list[0])
print("Last name :", name_list[-1])
print("elements from index 2:4 :", name_list[2:5])
print("Reverse :", name_list[::-1])
```

### DICTIONARY

```python
##(Creation, Modification and Access):

## CREATION 
student_mark = {"Nivedya": 56, "Avinash":86, "Taniya":67, "Yami":98, "Daya":83}
print(type(student_mark))
print(student_mark)
print("Mark:",student_mark["Yami"])

## UPDATE
student_mark.update({"Janani": 80})
print(student_mark)

student_mark.update({"Nivedya": 82})
print(student_mark)

students_mark = {"Name":"Nivedya", "Mark":82, "Name":"Avinash", "Mark":86, "Name":"Taniya", "Mark":67, "Name":"Yami", "Mark":98, "Name":"Daya", "Mark":33, "Name":"Janani", "Mark":80}
## KEY
print("keys :",students_mark.keys())

## VALUES
print("Values :", students_mark.values())

## ITEMS
print("Student name and marks :", students_mark.items())

```
