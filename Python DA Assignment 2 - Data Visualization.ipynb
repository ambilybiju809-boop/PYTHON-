### Loading the Taxis Dataset & Handling Missing Values

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Load the 'taxis' dataset
df = sns.load_dataset("taxis")
df
```

```python
df.info()
```

```python
df.describe()
```

#### df.isnull().sum()

```python
    ## filling the 7h
df['payment'] = df['payment'].fillna(df['payment'].mode()[0])
df
```

```python
## filling the missing values 
columns = ['pickup_zone','dropoff_zone','pickup_borough','dropoff_borough']
for column in columns:
    df[column] = df[column].fillna(df[column].mode()[0])
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df
```

```python
df.shape
```

```python
df.columns = df.columns.str.upper()
df
```

### Visualizations using Matplotlib/Pandas Plot:

```python
## line chart 
plt.figure(figsize=(12,6))
df = df.sort_values('PICKUP')
plt.plot(df['PICKUP'],df['FARE'])
plt.title("FARE OVER TIME")
plt.xlabel('PICKUP')
plt.ylabel('FARE')
plt.grid(True)
plt.show()
```

```python
## bar chart 
borough_fare = df.groupby('PICKUP_BOROUGH', observed=True)['FARE'].sum()
borough_fare.plot(kind='bar', figsize=(8, 5))
plt.title('TOTAL FARE BY PICKUP_BOROUGH')
plt.xlabel('PICKUP_BOROUGH')
plt.ylabel('FARE')
plt.show()
```

```python
## pie chart 
payment_counts = df['PAYMENT'].value_counts()
# Create pie chart
plt.figure(figsize=(7, 7))
plt.pie(payment_counts,
        labels=payment_counts.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Distribution Of Trips By Payment Method")
plt.show()
```

```python
## histogram 
plt.hist(df["DISTANCE"],bins = 20, edgecolor ='black')
plt.title(" Distribution Of Taxi Trip Distance")
plt.xlabel("Distance")
plt.ylabel("Frequency")
plt.show()
```

```python
## box plot
df.boxplot(column='TIP', by='PICKUP_BOROUGH')

plt.title('Distribution of Tips by Pickup Borough')
plt.suptitle('')
plt.xlabel('PICKUP_BOROUGH')
plt.ylabel('TIP AMOUNT')


plt.show()
```

### Visualizations Using Seaborn

```python
## count plot 
sns.countplot(data=df, x='PICKUP_BOROUGH')

plt.title('Number of Trips by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Number of Trips')
plt.show()
```

```python
## scatter plot 
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df,
                x='DISTANCE',
                y='FARE',
                hue='PICKUP_BOROUGH')
plt.title('Relationship Between Distance and Fare')
plt.xlabel('Distance')
plt.ylabel('Fare')
plt.show()
```
