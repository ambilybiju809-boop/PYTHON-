##### Numpy Array Operations:

```python
import numpy as np
```

```python
## 1. Create a 1D NumPy Array:
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(type(temperatures_w1))
print(temperatures_w1)
```

```python
## 2. Inspection and Properties:
print("Shape : ",temperatures_w1.shape)
print("Datatype :",temperatures_w1.dtype)
print("Size : ",temperatures_w1.size)
```

```python
## 3. Array Operations:
temperature_fahrenheit = (temperatures_w1 * 9/5) + 32
print("temperature in fahrenhiet ;", temperature_fahrenheit )
print("Maximum_temp ;",np.max(temperature_fahrenheit))
print("Mininmum_temp ;", np.min(temperature_fahrenheit))
print("Mean_temp ;", np.mean(temperature_fahrenheit))
```

```python
## 4. Array Slicing and Indexing:
print("First three days:", temperatures_w1[:3])
print("Weekend temperatures:", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

```python
## 5. Create a 2D Array:
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

```python
## 6. Inspect and Slice the 2D Array:
print("Shape ;", temperatures.shape)
print("Datatype ;" , temperatures.dtype)
print("Size : ",temperatures.size)
print("Week 1 temperatures:", temperatures[0, :])
print("Week 2 temperatures:", temperatures[1, :])
print("Weekend temperatures:", temperatures[:, -2:])
```

##### Pandas Series:

```python
import pandas as pd
```

```python
## 1. Creating Pandas Series:
marks = pd.Series([95, 92, 89, 85, 80], index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

```python
## 2. Indexing and Slicing:
print("1st Rank Mark:",marks.iloc[0])
print("Top 3 Ranks:", marks.loc[['Rank1', 'Rank2', 'Rank3']])
print("3rd Rank Mark:",marks.iloc[2])
print("Ranks with marks greater than 90:",marks[marks > 90])
```

```python
## 3. Manipulating Series:
marks.loc['Rank1'] = 100
print("Updated Marks:",marks)
marks = marks.drop('Rank5')
print("After Removing Last Rank:",marks)
cgpa = marks / 10
print("CGPA:",cgpa)
```

##### Pandas Dataframe:

```python
## 1. Creating Pandas DataFrame:
transaction =  {"TransactionID" : [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
                            "ProductCategory" : ['Electronics', 'Clothing', 'Electronics', 'Furniture','Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
                            "Region": ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South', 'North' ],
                            "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]}
transactions  = pd.DataFrame(transaction)
transactions
```

```python
## 2. Data Exploration:
print("Head :",transactions.head)
print("Tail :", transactions.tail)
print("Shape :", transactions.shape)
print("Column name:", transactions.columns)
print("Datatype :",transactions.dtypes)

```

```python
print("Names :",transactions[["ProductCategory","Amount"]])
print("Last 3 Columns:",transactions.iloc[:, -3:])
```

```python
print("North Region Transactions (Amount > 200):",transactions[(transactions['Region'] == 'North') &
                   (transactions['Amount'] > 200)])
print("Product Category Counts:",transactions['ProductCategory'].value_counts())
print("Unique Regions:",transactions['Region'].unique())
print("Mean Amount by Region:",transactions.groupby('Region')['Amount'].mean())

```

```python
## 3. Manipulating the DataFrame:
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("After Modifying Amount:")
print(transactions)

transactions['Discount'] = transactions['Amount'] * 0.10
print("After Adding Discount:",transactions)

transactions = transactions[transactions['TransactionID'] != 109]
print("After Removing TransactionID 109:",transactions)

transactions = transactions.drop('Discount', axis=1)
print("After Deleting Discount Column:",transactions)
```
